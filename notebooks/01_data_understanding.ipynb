```python
def missing_category(percent):
    if percent == 0:
        return "No Missing Values"
    elif percent < 5:
        return "Low (<5%)"
    elif percent < 30:
        return "Moderate (5-30%)"
    elif percent < 60:
        return "High (30-60%)"
    else:
        return "Very High (>60%)"


missing_summary = pd.DataFrame({
    "missing_percent": df.isnull().mean() * 100
})

missing_summary["category"] = (
    missing_summary["missing_percent"]
    .apply(missing_category)
)

print(
    missing_summary["category"]
    .value_counts()
)
```

```python
def cap_outliers(series, lower=0.01, upper=0.99):
    lower_bound = series.quantile(lower)
    upper_bound = series.quantile(upper)
    
    return series.clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)
```

```python
categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)
```

```python
print(preprocessor)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
baseline_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                class_weight="balanced",
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
```

```python
baseline_model.fit(X_train, y_train)
```

```python
from xgboost import XGBClassifier
```

```python
negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("Non-default:", negative_count)
print("Default:", positive_count)
print("scale_pos_weight:", scale_pos_weight)
```

```python
numeric_pipeline_xgb = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_pipeline_xgb = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocessor_xgb = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline_xgb, numeric_features),
        ("categorical", categorical_pipeline_xgb, categorical_features)
    ]
)
```

```python
xgb_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor_xgb),
        (
            "model",
            XGBClassifier(
                n_estimators=300,
                max_depth=6,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                scale_pos_weight=scale_pos_weight,
                objective="binary:logistic",
                eval_metric="logloss",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
print("Training XGBoost...")

xgb_model.fit(X_train, y_train)

print("XGBoost training completed!")
```

```python
xgb_y_pred = xgb_model.predict(X_test)

xgb_y_proba = xgb_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
```

```python
xgb_results = {
    "Accuracy": accuracy_score(y_test, xgb_y_pred),
    "Precision": precision_score(y_test, xgb_y_pred),
    "Recall": recall_score(y_test, xgb_y_pred),
    "F1": f1_score(y_test, xgb_y_pred),
    "ROC-AUC": roc_auc_score(y_test, xgb_y_proba),
    "PR-AUC": average_precision_score(y_test, xgb_y_proba)
}

xgb_results
```

```python
print(
    classification_report(
        y_test,
        xgb_y_pred,
        target_names=["Non-default", "Default"]
    )
)
```

```python
cm_xgb = confusion_matrix(y_test, xgb_y_pred)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm_xgb,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Predicted Non-default", "Predicted Default"],
    yticklabels=["Actual Non-default", "Actual Default"]
)

plt.title("XGBoost — Confusion Matrix")
plt.xlabel("Prediction")
plt.ylabel("Actual")
plt.show()
```

```python
model_comparison = pd.DataFrame([
    {"Model": "Logistic Regression", **baseline_results},
    {"Model": "XGBoost", **xgb_results}
])

model_comparison = model_comparison.set_index("Model")

model_comparison.round(4)
```

```python
def calculate_risk_score(pd_probability):
    pd_probability = np.asarray(pd_probability, dtype=float)
    pd_probability = np.clip(pd_probability, 0, 1)

    score = 850 - (550 * pd_probability)

    return np.round(np.clip(score, 300, 850)).astype(int)
```

```python
def categorize_risk(pd_probability):
    if pd_probability < 0.05:
        return "Low Risk"
    elif pd_probability < 0.15:
        return "Moderate Risk"
    elif pd_probability < 0.30:
        return "High Risk"
    else:
        return "Very High Risk"
```

```python
risk_scores = calculate_risk_score(calibrated_y_proba)

plt.figure(figsize=(10, 5))
sns.histplot(risk_scores, bins=30, kde=True)

plt.title("Calibrated Model — Credit Risk Score Distribution")
plt.xlabel("Project Risk Score (300–850)")
plt.ylabel("Number of Applicants")
plt.tight_layout()
plt.show()
```

```python
def make_loan_decision(pd_probability, review_threshold=0.10, reject_threshold=0.25):
    if not 0 <= pd_probability <= 1:
        raise ValueError("PD probability must be between 0 and 1.")

    if not 0 <= review_threshold < reject_threshold <= 1:
        raise ValueError(
            "Thresholds must satisfy 0 <= review_threshold < reject_threshold <= 1."
        )

    if pd_probability < review_threshold:
        return "Approve"
    elif pd_probability < reject_threshold:
        return "Manual Review"
    else:
        return "Reject"
```

```python
test_probabilities = [0.02, 0.12, 0.35]

for probability in test_probabilities:
    print(
        f"PD={probability:.0%} | "
        f"Score={calculate_risk_score([probability])[0]} | "
        f"Decision={make_loan_decision(probability)}"
    )
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix

threshold_rows = []

thresholds = np.arange(0.05, 0.51, 0.05)

for threshold in thresholds:
    predictions = (calibrated_y_proba >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test, predictions, labels=[0, 1]
    ).ravel()

    threshold_rows.append({
        "Threshold": round(threshold, 2),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1": f1_score(
            y_test, predictions, zero_division=0
        ),
        "False_Positives": fp,
        "False_Negatives": fn,
        "True_Positives": tp,
        "True_Negatives": tn
    })

threshold_analysis = pd.DataFrame(threshold_rows)
display(threshold_analysis.round(3))
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    threshold_analysis["Threshold"],
    threshold_analysis["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    threshold_analysis["Threshold"],
    threshold_analysis["Recall"],
    marker="o",
    label="Recall"
)

plt.plot(
    threshold_analysis["Threshold"],
    threshold_analysis["F1"],
    marker="o",
    label="F1"
)

plt.xlabel("Probability Threshold")
plt.ylabel("Metric")
plt.title("Threshold Analysis — Calibrated XGBoost")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
FN_COST = 5
FP_COST = 1

threshold_analysis["Illustrative_Cost"] = (
    threshold_analysis["False_Negatives"] * FN_COST
    + threshold_analysis["False_Positives"] * FP_COST
)

display(
    threshold_analysis[
        ["Threshold", "False_Positives", "False_Negatives",
         "Illustrative_Cost"]
    ].sort_values("Illustrative_Cost")
)
```
